## BIVARIATE NUMERIC PLOTS

```python
# using seaborn library and function relplot() [replative plot or replationship plot]
```

```python
sns.relplot('Sepal.Length','Petal.Length',data= iris);
```

```python
#there is a positive relation among the variables cause as petal length increases and sepal length also increases
```

## AREA PLOT

```python
iris.loc[:,['Sepal.Length','Petal.Length']].plot(kind='area')#All rows of the sepal length and petal length columns by names
```

```python
iris.iloc[:,[0,2]].plot(kind='area')#All rows of the sepal length and petal length columns by indexing
```

```python
#sepal length is stable, compared to petal length as petal length has more fluctuations 
#both sepal length and petal length are increasing i.e., positively correlated
```
